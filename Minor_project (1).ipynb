```python
import torch
from retinaface import RetinaFace
import cv2
import matplotlib.pyplot as plt


def detect_faces(image):
    faces = RetinaFace.detect_faces(image)
    return faces

# Detect faces in the drowsy and not drowsy images
faces_drowsy = detect_faces(drowsy_image)
faces_not_drowsy = detect_faces(not_drowsy_image)

def draw_faces(image, faces):
    for face in faces.values():
        # Get the bounding box coordinates
        facial_area = face['facial_area']
        # Draw a rectangle around the face
        cv2.rectangle(image,
                      (facial_area[0], facial_area[1]),
                      (facial_area[2], facial_area[3]),
                      (0, 255, 0), 2)
    return image

# Draw faces on the images
drowsy_image_with_faces = draw_faces(drowsy_image.copy(), faces_drowsy)
not_drowsy_image_with_faces = draw_faces(not_drowsy_image.copy(), faces_not_drowsy)

# Convert BGR to RGB for display
drowsy_image_rgb = cv2.cvtColor(drowsy_image_with_faces, cv2.COLOR_BGR2RGB)
not_drowsy_image_rgb = cv2.cvtColor(not_drowsy_image_with_faces, cv2.COLOR_BGR2RGB)

# Display the images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(drowsy_image_rgb)
plt.title("Drowsy with Face Detection")

plt.subplot(1, 2, 2)
plt.imshow(not_drowsy_image_rgb)
plt.title("Not Drowsy with Face Detection")

plt.show()
```

```python
import dlib
import cv2
import matplotlib.pyplot as plt

# Load the pre-trained shape predictor model
predictor_path = "shape_predictor_68_face_landmarks.dat"
detector = dlib.get_frontal_face_detector()
predictor = dlib.shape_predictor(predictor_path)

def get_landmarks(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    faces = detector(gray)
    landmarks = []
    for face in faces:
        landmarks.append(predictor(gray, face))
    return landmarks

def draw_landmarks(image, landmarks):
    for landmark in landmarks:
        for n in range(0, 68):
            x = landmark.part(n).x
            y = landmark.part(n).y
            cv2.circle(image, (x, y), 2, (0, 255, 0), 1)
    return image

# Get landmarks for drowsy and not drowsy images
landmarks_drowsy = get_landmarks(drowsy_image)
landmarks_not_drowsy = get_landmarks(not_drowsy_image)

# Draw landmarks on the images
drowsy_image_with_landmarks = draw_landmarks(drowsy_image.copy(), landmarks_drowsy)
not_drowsy_image_with_landmarks = draw_landmarks(not_drowsy_image.copy(), landmarks_not_drowsy)

# Convert BGR to RGB for display
drowsy_image_rgb = cv2.cvtColor(drowsy_image_with_landmarks, cv2.COLOR_BGR2RGB)
not_drowsy_image_rgb = cv2.cvtColor(not_drowsy_image_with_landmarks, cv2.COLOR_BGR2RGB)

# Display the images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(drowsy_image_rgb)
plt.title("Drowsy with Landmarks")

plt.subplot(1, 2, 2)
plt.imshow(not_drowsy_image_rgb)
plt.title("Not Drowsy with Landmarks")

plt.show()
```

```python
import cv2
import mediapipe as mp
import numpy as np
import matplotlib.pyplot as plt
import os

# Initialize MediaPipe Face Mesh
mp_face_mesh = mp.solutions.face_mesh
face_mesh = mp_face_mesh.FaceMesh(max_num_faces=1)
mp_drawing = mp.solutions.drawing_utils



# Load a specific image from a category
def load_image_from_file(file_name):
    image_path = os.path.join(dataset_path, file_name)
    image = cv2.imread(image_path)

    if image is None:
        print(f"Failed to load image: {image_path}")

    return image

# Estimate head pose from facial landmarks
def estimate_head_pose(landmarks):
    # Example: Compute the centroid of eyes and mouth corners
    eye_left = np.array([landmarks[33].x, landmarks[33].y])
    eye_right = np.array([landmarks[263].x, landmarks[263].y])
    nose_tip = np.array([landmarks[1].x, landmarks[1].y])
    mouth_left = np.array([landmarks[61].x, landmarks[61].y])
    mouth_right = np.array([landmarks[291].x, landmarks[291].y])

    # Compute simple relative distances for head pose estimation
    eye_distance = np.linalg.norm(eye_left - eye_right)
    nose_to_eye_distance = np.linalg.norm(nose_tip - eye_left)

    # For simplicity, we'll use these distances to estimate head pose
    yaw = np.arctan2(nose_tip[1] - (eye_left[1] + eye_right[1]) / 2, nose_tip[0] - (eye_left[0] + eye_right[0]) / 2)
    pitch = np.arctan2(eye_distance, nose_to_eye_distance)
    roll = 0  # Roll estimation requires more complex calculations

    return yaw, pitch, roll

# Process an image and estimate head pose
def process_image(image):
    if image is None:
        return None, None, None, None

    rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    results = face_mesh.process(rgb_image)

    if results.multi_face_landmarks:
        landmarks = results.multi_face_landmarks[0].landmark
        yaw, pitch, roll = estimate_head_pose(landmarks)
        return yaw, pitch, roll, results.multi_face_landmarks[0].landmark
    return None, None, None, None

# Draw head pose on the image
def draw_head_pose(image, landmarks, yaw, pitch, roll, scale=50):
    h, w = image.shape[:2]
    center = (w // 2, h // 2)

    # Draw facial landmarks
    for landmark in landmarks:
        x = int(landmark.x * w)
        y = int(landmark.y * h)
        cv2.circle(image, (x, y), 2, (0, 255, 0), -1)

    # Draw head pose (example with yaw and pitch)
    yaw = np.deg2rad(yaw)
    pitch = np.deg2rad(pitch)
    roll = np.deg2rad(roll)

    # Example lines for yaw and pitch
    cv2.line(image, center,
             (int(center[0] + scale * np.sin(pitch)), int(center[1] - scale * np.cos(pitch))),
             (0, 0, 255), 2)  # Pitch in red

    cv2.line(image, center,
             (int(center[0] + scale * np.sin(yaw)), int(center[1] - scale * np.cos(yaw))),
             (0, 255, 0), 2)  # Yaw in green

    return image

# Determine head position based on pitch
def determine_head_position(pitch):
    if pitch > 0.1:  # Threshold to determine "up" tilt
        return "Head is tilted up"
    elif pitch < -0.1:  # Threshold to determine "down" tilt
        return "Head is tilted down"
    else:
        return "Head is level"

# Load and process specific images
drowsy_image = load_image_from_file(r'/content/nthuddd2/train_data/drowsy/001_glasses_sleepyCombination_1289_drowsy.jpg')  # Replace with your drowsy image filename
not_drowsy_image = load_image_from_file(r'/content/nthuddd2/train_data/notdrowsy/001_glasses_nonsleepyCombination_1899_notdrowsy.jpg')  # Replace with your not drowsy image filename

if drowsy_image is not None:
    yaw_drowsy, pitch_drowsy, roll_drowsy, landmarks_drowsy = process_image(drowsy_image)
if not_drowsy_image is not None:
    yaw_not_drowsy, pitch_not_drowsy, roll_not_drowsy, landmarks_not_drowsy = process_image(not_drowsy_image)

# Ensure landmarks are found before drawing
if landmarks_drowsy:
    drowsy_image_with_pose = draw_head_pose(drowsy_image.copy(), landmarks_drowsy, yaw_drowsy, pitch_drowsy, roll_drowsy)
    head_position_drowsy = determine_head_position(pitch_drowsy)
if landmarks_not_drowsy:
    not_drowsy_image_with_pose = draw_head_pose(not_drowsy_image.copy(), landmarks_not_drowsy, yaw_not_drowsy, pitch_not_drowsy, roll_not_drowsy)
    head_position_not_drowsy = determine_head_position(pitch_not_drowsy)

# Convert BGR to RGB for display
if drowsy_image_with_pose is not None:
    drowsy_image_rgb = cv2.cvtColor(drowsy_image_with_pose, cv2.COLOR_BGR2RGB)
if not_drowsy_image_with_pose is not None:
    not_drowsy_image_rgb = cv2.cvtColor(not_drowsy_image_with_pose, cv2.COLOR_BGR2RGB)

# Display the images with head position information
plt.figure(figsize=(10, 5))

if drowsy_image_with_pose is not None:
    plt.subplot(1, 2, 1)
    plt.imshow(drowsy_image_rgb)
    plt.title(f"Drowsy: {head_position_drowsy}")

if not_drowsy_image_with_pose is not None:
    plt.subplot(1, 2, 2)
    plt.imshow(not_drowsy_image_rgb)
    plt.title(f"Not Drowsy: {head_position_not_drowsy}")

plt.show()



```

```python
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
import torch.nn.functional as F

class HybridDrowsinessModel(nn.Module):
    def __init__(self):
        super(HybridDrowsinessModel, self).__init__()
        # CNN layers
        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, stride=1, padding=1)
        self.conv2 = nn.Conv2d(16, 32, kernel_size=3, stride=1, padding=1)
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2, padding=0)

        # LSTM layers
        self.lstm = nn.LSTM(32 * 32 * 32, 128, batch_first=True)

        # Fully connected layers
        self.fc1 = nn.Linear(128, 64)
        self.fc2 = nn.Linear(64, 2)

    def forward(self, x):
        # CNN forward pass
        batch_size, sequence_len, c, h, w = x.size()
        cnn_out = []
        for t in range(sequence_len):
            x_t = x[:, t, :, :, :]
            x_t = self.pool(F.relu(self.conv1(x_t)))
            x_t = self.pool(F.relu(self.conv2(x_t)))
            cnn_out.append(x_t.view(batch_size, -1))

        cnn_out = torch.stack(cnn_out, dim=1)

        # LSTM forward pass
        lstm_out, _ = self.lstm(cnn_out)
        lstm_out = lstm_out[:, -1, :]  # Get the output from the last LSTM cell

        # Fully connected layers
        x = F.relu(self.fc1(lstm_out))
        x = self.fc2(x)

        return x

# Instantiate the hybrid model
model = HybridDrowsinessModel()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torchvision import datasets, transforms
from torch.utils.data import DataLoader, random_split, Dataset
from PIL import Image
import os
import time
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score, roc_curve, auc, precision_recall_curve

# Define transformations
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

# Custom dataset class (optional if using ImageFolder)
class CustomDataset(Dataset):
    def __init__(self, data_dir, transform=None):
        self.data_dir = data_dir
        self.transform = transform
        self.image_files = [f for f in os.listdir(data_dir) if f.endswith('.jpg')]
        self.labels = [0 if 'drowsy' in f else 1 for f in self.image_files]  # Binary classification

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_name = os.path.join(self.data_dir, self.image_files[idx])
        image = Image.open(img_name)
        label = self.labels[idx]

        if self.transform:
            image = self.transform(image)

        return image, label

# Use ImageFolder to load the dataset
dataset = datasets.ImageFolder(root='/content/nthuddd2/train_data', transform=transform)

total_size = len(dataset)
train_size = int(0.8 * total_size)
test_size = total_size - train_size

train_dataset, test_dataset = random_split(dataset, [train_size, test_size])

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

# Simple CNN model definition
class SimpleCNN(nn.Module):
    def __init__(self):
        super(SimpleCNN, self).__init__()
        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, stride=1, padding=1)
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2, padding=0)
        self.fc1 = nn.Linear(16 * 112 * 112, 2)  # Adjust based on image size and number of classes

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))
        x = x.view(-1, 16 * 112 * 112)
        x = self.fc1(x)
        return x

# Initialize the model, criterion, and optimizer
model = SimpleCNN()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Training loop with both training and testing loss tracking
def train_model(model, train_loader, test_loader, criterion, optimizer, num_epochs=10):
    model.train()
    train_losses = []
    test_losses = []
    start_time = time.time()  # Start timer for training

    for epoch in range(num_epochs):
        running_loss = 0.0
        for inputs, labels in train_loader:
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item()

        # Calculate the average training loss for this epoch
        epoch_train_loss = running_loss / len(train_loader)
        train_losses.append(epoch_train_loss)

        # Calculate the test loss after each epoch
        model.eval()
        test_loss = 0.0
        with torch.no_grad():
            for inputs, labels in test_loader:
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                test_loss += loss.item()
        epoch_test_loss = test_loss / len(test_loader)
        test_losses.append(epoch_test_loss)

        model.train()  # Set back to train mode

        print(f"Epoch {epoch+1}/{num_epochs}, Training Loss: {epoch_train_loss:.4f}, Testing Loss: {epoch_test_loss:.4f}")

    end_time = time.time()  # End timer for training
    training_time = end_time - start_time
    print(f"Training Time: {training_time:.2f} seconds")

    return train_losses, test_losses, training_time

# Train the model and track both training and testing losses
train_losses, test_losses, training_time = train_model(model, train_loader, test_loader, criterion, optimizer)

# Evaluate the model and collect metrics
def evaluate_model(model, test_loader):
    model.eval()
    all_labels = []
    all_preds = []
    all_scores = []
    start_time = time.time()  # Start timer for testing

    with torch.no_grad():
        for inputs, labels in test_loader:
            outputs = model(inputs)
            _, predicted = torch.max(outputs.data, 1)
            all_labels.extend(labels.cpu().numpy())
            all_preds.extend(predicted.cpu().numpy())
            all_scores.extend(F.softmax(outputs, dim=1).cpu().numpy())  # For ROC and PR curves

    end_time = time.time()  # End timer for testing
    testing_time = end_time - start_time

    accuracy = 100 * sum(np.array(all_preds) == np.array(all_labels)) / len(all_labels)
    precision = precision_score(all_labels, all_preds)
    recall = recall_score(all_labels, all_preds)
    f1 = f1_score(all_labels, all_preds)

    print(f"Accuracy: {accuracy:.2f}%")
    print(f"Precision: {precision:.2f}")
    print(f"Recall: {recall:.2f}")
    print(f"F1 Score: {f1:.2f}")
    print(f"Testing Time: {testing_time:.2f} seconds")

    return all_labels, all_preds, np.array(all_scores), testing_time

# Evaluate the model
all_labels, all_preds, all_scores, testing_time = evaluate_model(model, test_loader)

# Plot Training and Testing Losses
def plot_losses(train_losses, test_losses):
    plt.figure(figsize=(10, 6))
    plt.plot(train_losses, label='Training Loss')
    plt.plot(test_losses, label='Testing Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.title('Training and Testing Loss Over Epochs')
    plt.legend()
    plt.show()

# Plot Confusion Matrix
def plot_confusion_matrix(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
    plt.title('Confusion Matrix')
    plt.xlabel('Predicted Labels')
    plt.ylabel('True Labels')
    plt.show()

# Plot ROC Curve
def plot_roc_curve(y_true, y_scores):
    fpr, tpr, _ = roc_curve(y_true, y_scores[:, 1])
    roc_auc = auc(fpr, tpr)
    plt.figure()
    plt.plot(fpr, tpr, label=f'ROC Curve (AUC = {roc_auc:.2f})')
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC Curve')
    plt.legend(loc='lower right')
    plt.show()

# Plot Precision-Recall Curve
def plot_precision_recall_curve(y_true, y_scores):
    precision, recall, _ = precision_recall_curve(y_true, y_scores[:, 1])
    plt.figure()
    plt.plot(recall, precision, label='Precision-Recall Curve')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title('Precision-Recall Curve')
    plt.legend(loc='lower left')
    plt.show()

# Plot the loss curves
plot_losses(train_losses, test_losses)

# Plot the confusion matrix, ROC curve, and precision-recall curve
plot_confusion_matrix(all_labels, all_preds)
plot_roc_curve(np.array(all_labels), all_scores)
plot_precision_recall_curve(np.array(all_labels), all_scores)

```
